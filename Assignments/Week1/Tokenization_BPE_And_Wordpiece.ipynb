```python
# Byte Pair Encoding (BPE) - General Python Program

# Input words with frequencies
word_freq = {
    "Low": 5,
    "Lower": 2,
    "Newest": 3,
    "Wildest": 4
}

# Convert words into character tokens
tokens = {word: list(word.lower()) for word in word_freq}

# Function to calculate adjacent pair frequencies
def calculate_pair_freq(tokens, word_freq):
    pair_freq = {}

    for word, token_list in tokens.items():
        freq = word_freq[word]

        for i in range(len(token_list) - 1):
            pair = (token_list[i], token_list[i + 1])
            pair_freq[pair] = pair_freq.get(pair, 0) + freq

    return pair_freq

# Function to merge the most frequent pair
def merge_pair(tokens, pair):
    new_tokens = {}

    for word, token_list in tokens.items():
        merged = []
        i = 0

        while i < len(token_list):
            if (
                i < len(token_list) - 1
                and token_list[i] == pair[0]
                and token_list[i + 1] == pair[1]
            ):
                merged.append(pair[0] + pair[1])
                i += 2
            else:
                merged.append(token_list[i])
                i += 1

        new_tokens[word] = merged

    return new_tokens

# Number of merging iterations
num_iterations = 10

# Perform BPE
for iteration in range(num_iterations):

    pair_freq = calculate_pair_freq(tokens, word_freq)

    if not pair_freq:
        break

    # Find the most frequent pair
    best_pair = max(pair_freq, key=pair_freq.get)

    print(f"\nIteration {iteration + 1}")
    print("Pair Frequencies:", pair_freq)
    print("Selected Pair:", best_pair)
    print("Frequency:", pair_freq[best_pair])

    # Merge the selected pair
    tokens = merge_pair(tokens, best_pair)

    # Display updated tokens
    print("Updated Tokens:")
    for word, token_list in tokens.items():
        print(word, ":", token_list)

# Final tokens
print("\nFinal Tokenization:")
for word, token_list in tokens.items():
    print(word, ":", token_list)
```
