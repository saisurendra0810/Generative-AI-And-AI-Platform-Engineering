```python
!pip install tiktoken

import tiktoken

# Tokenizer
tokenizer = tiktoken.get_encoding("cl100k_base")

# Original prompt
original_prompt = """Can you please explain the concept of machine learning
in a very simple and easy-to-understand way? I want you to explain it
as if I am a beginner who has no prior knowledge about this topic.
Please also provide one simple real-world example."""

# Optimized prompt
optimized_prompt = """Explain machine learning simply for a beginner
with one real-world example."""

# Count tokens
original_tokens = len(tokenizer.encode(original_prompt))
optimized_tokens = len(tokenizer.encode(optimized_prompt))

# Calculate savings
saved_tokens = original_tokens - optimized_tokens
reduction = (saved_tokens / original_tokens) * 100

# Display results
print("Original Prompt:", original_prompt)
print("\nOptimized Prompt:", optimized_prompt)
print("\nOriginal Tokens:", original_tokens)
print("Optimized Tokens:", optimized_tokens)
print("Tokens Saved:", saved_tokens)
print("Token Reduction:", round(reduction, 2), "%")
```
