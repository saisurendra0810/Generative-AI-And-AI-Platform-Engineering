```python

import numpy as np

# =====================================
# STEP 1: Define Query Vector
# =====================================

# Query vector for the word "Love"
Q = np.array([1, 1])

print("STEP 1: Query Vector")
print("Q =", Q)


# =====================================
# STEP 2: Define Key Vectors
# =====================================

keys = {
    "I": np.array([1, 0]),
    "Love": np.array([1, 1]),
    "AI": np.array([0, 1])
}

print("\nSTEP 2: Key Vectors")
for word, key in keys.items():
    print(word, "=", key)


# =====================================
# STEP 3: Calculate Similarity Scores
# =====================================

scores = []

print("\nSTEP 3: Similarity Scores")

for word, key in keys.items():
    score = np.dot(Q, key)
    scores.append(score)
    print(f"{word}: {Q} . {key} = {score}")

scores = np.array(scores)

print("Similarity Scores:", scores)


# =====================================
# STEP 4: Apply Softmax
# =====================================

print("\nSTEP 4: Apply Softmax")

exp_scores = np.exp(scores)

print("Exponentials:", exp_scores)

denominator = np.sum(exp_scores)

print("Denominator:", denominator)

softmax = exp_scores / denominator

print("Softmax Values:", softmax)


# =====================================
# STEP 5: Calculate Attention Weights
# =====================================

print("\nSTEP 5: Attention Weights")

for word, weight in zip(keys.keys(), softmax):
    print(f"{word}: {weight:.3f} ({weight * 100:.1f}%)")

print("\nFinal Attention Weights:", np.round(softmax, 3))
print("Total Attention:", np.sum(softmax))
```
