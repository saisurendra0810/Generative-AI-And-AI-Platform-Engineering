```python
import time
import uuid
from datetime import datetime, timezone

# Gemini 3.5 Flash Standard paid-tier pricing
INPUT_PRICE = 1.50
OUTPUT_PRICE = 9.00

def ask_gemini(prompt, max_attempts=3):

    request_id = str(uuid.uuid4())

    for attempt in range(1, max_attempts + 1):

        start_time = time.perf_counter()

        record = {
            "_id": f"{request_id}_{attempt}",
            "request_id": request_id,
            "timestamp": datetime.now(timezone.utc),
            "model": MODEL_NAME,
            "prompt": prompt,
            "attempt": attempt,
            "input_tokens": 0,
            "output_tokens": 0,
            "total_tokens": 0,
            "latency_seconds": 0,
            "estimated_cost_usd": 0,
            "status": "failed",
            "error": None
        }

        try:
            print(f"Attempt {attempt}...")

            response = client.models.generate_content(
                model=MODEL_NAME,
                contents=prompt
            )

            usage = response.usage_metadata

            input_tokens = getattr(usage, "prompt_token_count", 0) or 0
            output_tokens = getattr(usage, "candidates_token_count", 0) or 0
            total_tokens = getattr(usage, "total_token_count", 0) or 0

            latency = time.perf_counter() - start_time

            estimated_cost = (
                (input_tokens * INPUT_PRICE / 1_000_000) +
                (output_tokens * OUTPUT_PRICE / 1_000_000)
            )

            record.update({
                "input_tokens": input_tokens,
                "output_tokens": output_tokens,
                "total_tokens": total_tokens,
                "latency_seconds": latency,
                "estimated_cost_usd": estimated_cost,
                "status": "success"
            })

            collection.insert_one(record)

            print("\nGemini Response:")
            print(response.text)

            print("\nRequest Details:")
            print("Input Tokens:", input_tokens)
            print("Output Tokens:", output_tokens)
            print("Total Tokens:", total_tokens)
            print("Latency:", round(latency, 3), "seconds")
            print("Estimated Cost: $", round(estimated_cost, 8))

            return response.text

        except Exception as e:

            record["latency_seconds"] = time.perf_counter() - start_time
            record["error"] = str(e)[:1000]

            collection.insert_one(record)

            print("Error:", str(e)[:300])

            if attempt < max_attempts:
                wait_time = attempt * 10
                print(f"Retrying in {wait_time} seconds...")
                time.sleep(wait_time)

            else:
                print("All attempts failed.")
                return None
```

```python
import pandas as pd

records = list(
    collection.find().sort("timestamp", -1).limit(10)
)

df = pd.DataFrame(records)

print("Total records retrieved:", len(df))

display(df[[
    "model",
    "attempt",
    "input_tokens",
    "output_tokens",
    "total_tokens",
    "latency_seconds",
    "estimated_cost_usd",
    "status"
]])
```

```python
pipeline = [
    {
        "$match": {
            "status": "success"
        }
    },
    {
        "$group": {
            "_id": None,
            "total_requests": {"$sum": 1},
            "total_input_tokens": {"$sum": "$input_tokens"},
            "total_output_tokens": {"$sum": "$output_tokens"},
            "total_tokens": {"$sum": "$total_tokens"},
            "average_latency": {"$avg": "$latency_seconds"},
            "total_estimated_cost": {"$sum": "$estimated_cost_usd"}
        }
    }
]

results = list(collection.aggregate(pipeline))

if results:
    stats = results[0]

    print("MONGODB USAGE ANALYTICS")
    print("----------------------")
    print("Total Requests:", stats["total_requests"])
    print("Total Input Tokens:", stats["total_input_tokens"])
    print("Total Output Tokens:", stats["total_output_tokens"])
    print("Total Tokens:", stats["total_tokens"])
    print("Average Latency:", round(stats["average_latency"], 3), "seconds")
    print("Total Estimated Cost: $", round(stats["total_estimated_cost"], 8))
else:
    print("No successful API records found.")
```

```python
import matplotlib.pyplot as plt

# Token usage graph
token_labels = ["Input Tokens", "Output Tokens"]
token_values = [
    stats["total_input_tokens"],
    stats["total_output_tokens"]
]

plt.figure(figsize=(8, 5))
plt.bar(token_labels, token_values)
plt.title("Gemini API Token Usage")
plt.ylabel("Number of Tokens")
plt.show()


# Latency graph
plt.figure(figsize=(8, 5))
plt.bar(
    ["Average Latency"],
    [stats["average_latency"]]
)
plt.title("Gemini API Average Latency")
plt.ylabel("Seconds")
plt.show()


# Estimated cost graph
plt.figure(figsize=(8, 5))
plt.bar(
    ["Estimated Cost"],
    [stats["total_estimated_cost"]]
)
plt.title("Gemini API Estimated Cost")
plt.ylabel("Cost (USD)")
plt.show()
```
