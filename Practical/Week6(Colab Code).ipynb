```python
import requests

OLLAMA_URL = "http://localhost:11434/api/generate"
MODEL_NAME = "llama3.2"

def get_llm_answer(question):
    payload = {
        "model": MODEL_NAME,
        "prompt": question,
        "stream": False,
        "options": {
            "temperature": 0,
            "num_predict": 150
        }
    }

    response = requests.post(
        OLLAMA_URL,
        json=payload,
        timeout=300
    )

    response.raise_for_status()

    return response.json()["response"].strip()

print("LLM function created successfully!")
```

```python
test_question = "What is tokenization in large language models?"

answer = get_llm_answer(test_question)

print("QUESTION:")
print(test_question)

print("\nLLAMA 3.2 ANSWER:")
print(answer)
```

```python
def calculate_score(answer, expected_keywords):

    answer = answer.lower()

    matched_keywords = []
    missing_keywords = []

    for keyword in expected_keywords:
        if keyword.lower() in answer:
            matched_keywords.append(keyword)
        else:
            missing_keywords.append(keyword)

    total_keywords = len(expected_keywords)

    if total_keywords == 0:
        score = 0
    else:
        score = (len(matched_keywords) / total_keywords) * 100

    return round(score, 2), matched_keywords, missing_keywords


print("Evaluation function created successfully!")
```

```python
sample_answer = "Tokenization breaks text into smaller tokens."

sample_keywords = ["text", "tokens"]

score, matched, missing = calculate_score(
    sample_answer,
    sample_keywords
)

print("Score:", score, "%")
print("Matched keywords:", matched)
print("Missing keywords:", missing)
```

```python
import pandas as pd
from datetime import datetime

results = []

for i, item in enumerate(golden_questions):

    question = item["question"]

    print(f"Evaluating question {i + 1}/30...")

    # Get answer from Llama 3.2
    answer = get_llm_answer(question)

    # Calculate keyword score
    score, matched, missing = calculate_score(
        answer,
        item["keywords"]
    )

    results.append({
        "id": item["id"],
        "question": question,
        "expected_answer": item["expected_answer"],
        "llm_answer": answer,
        "score": score,
        "matched_keywords": ", ".join(matched),
        "missing_keywords": ", ".join(missing),
        "timestamp": datetime.now().isoformat()
    })

print("\nEvaluation completed successfully!")
print("Total questions evaluated:", len(results))
```
